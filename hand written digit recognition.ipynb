```python
#digit range is b/w 0-255. We need to normalize to 0-1
#to ensure the vals to neural n/w are in same scale
#avoids the situation where huger no's dominate the traing process(thinking them as imp features)
#faster convergence(reaching the global min)
#smooth gradientt descent
#->normalization speeds up training(easy to calculate gradients,backpropagating..etc)
#->helps avoiding exploding weights
x_train_normalized=x_train/255
x_test_normalized=x_test/255
```
